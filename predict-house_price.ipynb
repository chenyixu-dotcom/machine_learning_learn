```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy= 'median')
```

```python
help(imputer.fit)
```

```python
numeric_col = train_df.select_dtypes(include='number').drop(columns= ['SalePrice', 'Id']).columns
train_target_df = train_df.SalePrice
imputer.fit(train_df[numeric_col])
train_df[numeric_col] = imputer.transform(train_df[numeric_col])
test_df[numeric_col] = imputer.transform(test_df[numeric_col])
```

```python
test_df[numeric_col].isna().any()
# check if na remain
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(train_df[numeric_col])
train_df[numeric_col] = scaler.transform(train_df[numeric_col])
test_df[numeric_col] = scaler.transform(test_df[numeric_col])
```

```python
# treat categorical data into numerical data via one hot encoder

from sklearn.preprocessing import OneHotEncoder
categorical_col = train_df.select_dtypes(include= 'object').columns

enc = OneHotEncoder(sparse_output= False, handle_unknown= 'ignore')
enc.fit(train_df[categorical_col])
```

```python
enc_col = enc.get_feature_names_out(categorical_col)
train_df[enc_col] = enc.transform(train_df[categorical_col])
test_df[enc_col] = enc.transform(test_df[categorical_col])
```

```python
numeric_col.to_list()
```

```python
enc_col
```

```python
from sklearn.linear_model import LinearRegression

col = numeric_col.tolist() + enc_col.tolist()
X_train = train_df[col]
X_test = test_df[col]
y = train_target_df

model = LinearRegression()
model.fit(X_train, y)

```

```python
from math import sqrt
def RMSE(predict, target):

    score = sqrt(((predict - target)**2).sum()/len(predict))
    return score
```

```python
predict = model.predict(X_train)
RMSE(predict, train_target_df)
# for house price averaging like 300k its not that bad
```

```python
# get baseline rmse to better estimate the accuracy of the model

min_price = train_target_df.min()
max_price = train_target_df.max()
median_price = train_target_df.median()

np.random.seed(42)
random_prediction = np.random.uniform(min_price, max_price, len(train_target_df))
median_prediction = np.full(len(train_target_df), median_price)

random_baseline = RMSE(random_prediction, train_target_df)
median_baseline = RMSE(median_prediction, train_target_df)
print(random_baseline, median_baseline)
```

```python
from sklearn.linear_model import Ridge

ridge_model = Ridge()
ridge_model.fit(X_train, y)
ridge_prediction = ridge_model.predict(X_train)
RMSE(ridge_prediction, train_target_df)
```
